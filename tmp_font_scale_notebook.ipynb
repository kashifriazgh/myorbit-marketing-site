```python
from pathlib import Path
import re

root = Path('app/dashboard/clients')
file_patterns = ['*.css', '*.module.css', '*.tsx', '*.ts', '*.jsx', '*.js']

# Collect files
files = []
for pattern in file_patterns:
    files.extend(sorted(root.rglob(pattern)))

print(f'Found {len(files)} target files:')
for f in files:
    print('-', f)

# Define replacement rules
css_replacements = [
    (r'font-size:\s*([0-9.]+)px;', lambda m: f'font-size: {round(float(m.group(1)) * 1.25, 4)}px;'),
]
jsx_replacements = [
    (r'\btext-5xl\b', 'text-6xl'),
    (r'\btext-4xl\b', 'text-5xl'),
    (r'\btext-3xl\b', 'text-4xl'),
    (r'\btext-2xl\b', 'text-3xl'),
    (r'\btext-xl\b', 'text-2xl'),
    (r'\btext-lg\b', 'text-xl'),
    (r'\btext-base\b', 'text-lg'),
    (r'\btext-sm\b', 'text-base'),
    (r'\btext-xs\b', 'text-sm'),
]

changed_files = []
for path in files:
    text = path.read_text(encoding='utf-8')
    orig = text
    if path.suffix == '.css':
        for pattern, repl in css_replacements:
            text = re.sub(pattern, repl, text)
    elif path.suffix in {'.tsx', '.ts', '.jsx', '.js'}:
        for pattern, repl in jsx_replacements:
            text = re.sub(pattern, repl, text)
    if text != orig:
        path.write_text(text, encoding='utf-8')
        changed_files.append(path)

print(f'Updated {len(changed_files)} files:')
for f in changed_files:
    print('-', f)
```
